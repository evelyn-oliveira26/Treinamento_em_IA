```python
from sklearn.datasets import load_iris
import pandas as pd
import matplotlib.pyplot as plt

iris = load_iris(as_frame=True)
df_iris = iris.frame

df_iris["species"] = df_iris["target"].map(
    dict(enumerate(iris.target_names))
)

print("Primeiras Linhas:\n", df_iris.head())
print("\nTamanho:", df_iris.shape)
print("\nColunas:\n", df_iris.columns)
print("\nInformações:", df_iris.info())
print("\nValores Ausentes:\n", df_iris.isnull().sum())
print("\nDuplicatas:", df_iris.duplicated().sum())
print("\nEstatísticas Descritivas:\n", df_iris.describe())

print("\nDistribuição das Espécies:")
print(df_iris["species"].value_counts().sort_index())

print("\nMédia de Comprimento da Sépala:")
print(df_iris.groupby("species")["sepal length (cm)"].mean())

#barras
plt.bar(df_iris["species"], df_iris["sepal length (cm)"])
plt.xlabel("Espécie")
plt.ylabel("Comprimento")
plt.title("Comprimento por Espécie")
plt.show()

#histograma
plt.hist(df_iris["sepal length (cm)"])
plt.xlabel("Comprimento")
plt.ylabel("Quantidade")
plt.title("Distribuição de Comprimento - Sépala")
plt.show()

#dispersão
plt.scatter(df_iris["petal width (cm)"], df_iris["petal length (cm)"])
plt.xlabel("Largura")
plt.ylabel("Comprimento")
plt.title("Largura x Comprimento - Pétala")
plt.show()

#boxplot
plt.boxplot(df_iris["petal width (cm)"])
plt.ylabel("Largura")
plt.title("Largura das Pétalas")
plt.show()

print(df_iris[["petal width (cm)", "sepal width (cm)"]].corr())
```

**Insight 1:** O dataset está equilibrado entre as três espécies, com 50 amostras para cada uma. Isso é relevante para uma classificação futura, porque nenhuma espécie está desproporcionalmente representada.

**Insight 2:** As espécies apresentam diferenças no comprimento médio da sépala, sendo a "Iris Virginica" a que apresenta a maior média e a "Iris Setosa" a menor. Isso mostra que o comprimento da sépala pode ser uma característica relevante para diferenciar as espécies e contribuir para o desempenho de um futuro modelo de classificação.

**Insight 3:** O dataset não apresenta valores ausentes em nenhuma das variáveis analisadas, o que significa que todas as 150 amostras possuem informações completas. Isso facilita a análise dos dados e evita a necessidade de realizar tratamentos ou preenchimentos antes de utilizá-los em um futuro modelo de classificação.

**Insight 4:** Existe uma correlação entre a largura da pétala e a largura da sépala, porém ela é relativamente fraca. Isso indica que essas duas características possuem uma relação limitada. No caso a largura da sépala, isoladamente, não explica bem as variações na largura da pétala.
